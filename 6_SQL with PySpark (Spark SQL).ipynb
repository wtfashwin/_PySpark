```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("SparkSQLExample").getOrCreate()

data = [
    (1, "Alice", "HR", 5000),
    (2, "Bob", "IT", 7000),
    (3, "Charlie", "Finance", 6500)
]

df = spark.createDataFrame(data, ["id", "name", "department", "salary"])
```

```python
df.show()
```

```python


# Register DataFrame as a SQL table
df.createOrReplaceTempView('employees')

# Run SQL Query
result = spark.sql("SELECT  * FROM employees")
result.show()
```

```python
spark.sql('SELECT * FROM employees where salary > 5000').show()
```

```python
spark.sql("SELECT name, salary FROM employees WHERE department = 'IT'").show()
```

SQL Functions in PySpark
PySpark provides built-in SQL functions for performing transformations.

Aggregate Functions

```python
spark.sql("""
    SELECT department, SUM(salary) AS total_salary, AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department
""").show()
```

### String Functions (concat(), substr(), length())

Helps in string manipulation like concatenation and substring extraction.

```python
#Concatenating Name and Department
spark.sql("SELECT name, department, concat(name, '-', department)as Details FROM employees ").show()
```

```python
#Extracting the First 3 Letters of Each Name
spark.sql("SELECT name, SUBSTRING(name, 1,3) as short_info from employees").show()
```

```python
#Extracting the First 3 Letters of Each Name
spark.sql("SELECT name, SUBSTRING(name, 1, 2) AS short_name FROM employees").show()
```

```python
#Finding the Length of Each Employee Name
spark.sql("SELECT name, LENGTH(name) as name_length from employees").show()
```

```python
from pyspark.sql import SparkSession

# Create Spark session
spark = SparkSession.builder.appName("SparkSQL").getOrCreate()

# Sample Employee Data
data = [("Amit", "IT", 50000), ("Priya", "HR", 60000), ("Ravi", "Finance", 55000)]
df = spark.createDataFrame(data, ["name", "department", "salary"])
df.show()
```

```python
# Register DataFrame as Table
df.createOrReplaceTempView("employees")
```

```python
# Query the table
spark.sql("SELECT * FROM employees").show()
```

### A. current_date() – Get Today’s Date

```python
# Example 1: Fetch Current Date
spark.sql("SELECT CURRENT_DATE() AS TODAY").show()
```

```python
# Example 2: Add a current_date column to employees
spark.sql("SELECT name, department, salary, current_date() as today from employees").show()
```

### B. datediff() – Find Difference Between Two Dates

```python
# Example 1: Find difference between two dates
spark.sql("SELECT datediff('2025-07-01','2025-05-01') as days_difference").show()
```

```python
#Example 2: Find days since employee joined

data = [("Amit", "2025-08-15"), ("Priya", "2025-07-10")]

df = spark.createDataFrame(data, ["name", "join_date"])
df.show()

```

```python
df.createOrReplaceTempView("emp_data")
```

```python
#Find days since employee joined

spark.sql("SELECT name, datediff(current_date(), join_date) AS days_worked from emp_data").show()
```

### C. date_add() and date_sub() – Add/Subtract Days

```python
# Example 1: Add 10 days to the current date

spark.sql("SELECT date_add(current_date(), 10) as future_date").show()
```

```python
# Example 2: Subtract 30 days from a given date

spark.sql("select date_sub('2025-09-10', 30) as previous_date").show()
```
