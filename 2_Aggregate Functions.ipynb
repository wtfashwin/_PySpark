### ✅ What are Aggregate Functions?
In Spark, aggregate functions are used to perform calculations on a group of rows and return a single result. These functions are typically used with groupBy() or agg() in Spark DataFrames.

### Aggregate functions with groupBy

```python
from pyspark.sql import functions as F
```

```python
#max()
df.groupBy('DEPT').agg(F.max('FEE')).show()
```

```python
#min()
df.groupBy('DEPT').agg(F.min('FEE')).show()
```

```python
#sum()
df.groupBy('DEPT').agg(F.sum('FEE')).show()
```

```python
#count()
df.groupBy('DEPT').agg(F.count('FEE')).show()
```

### Apply all the aggregate functions at one time

```python
df.groupBy('DEPT').agg(F.min('FEE'),
                       F.max('FEE'),
                       F.avg('FEE'),
                       F.sum('FEE'),
                       F.mean('FEE')).show()
```

### Apply groupBy on multiple column

```python
df.groupBy('DEPT','NAME').agg(F.max('FEE'),
                       F.min('FEE'),
                       F.avg('FEE'),
                       F.sum('FEE'),
                       F.count('FEE'),
                       F.mean('FEE')).show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, avg, max, min, count

# Create Spark session
spark = SparkSession.builder.master("local").appName("AggregateFunctions").getOrCreate()

# Sample Employee Data
data = [
    (101, "Alice", "HR", 50000, 30),
    (102, "Bob", "IT", 60000, 32),
    (103, "Charlie", "HR", 55000, 29),
    (104, "David", "IT", 75000, 35),
    (105, "Eve", "Finance", 80000, 40)
]

# Define schema and create DataFrame
columns = ["emp_id", "name", "department", "salary", "age"]
df = spark.createDataFrame(data, schema=columns)

# Display the DataFrame
df.show()

# 1. Count of employees in each department
df.groupBy("department").agg(count("*").alias("num_employees")).show()

# 2. Sum of salaries by department
df.groupBy("department").agg(sum("salary").alias("total_salary")).show()

# 3. Average salary by department
df.groupBy("department").agg(avg("salary").alias("avg_salary")).show()

# 4. Maximum salary in each department
df.groupBy("department").agg(max("salary").alias("max_salary")).show()

# 5. Minimum age in each department
df.groupBy("department").agg(min("age").alias("min_age")).show()

# Stop Spark session
spark.stop()
```
