### WithColumn
To add a new column to an existing DataFrame in PySpark, you can use the withColumn() method. This method allows you to create a new column or replace an existing one by applying a transformation.

```python
#Syntax
df = df.withColumn("new_column_name", expression)

# new_column_name is the name of the new column.
# expression is the transformation or value that will populate the new column.
```

### Adding a Column Using a Conditional Expression

```python
from pyspark.sql import SparkSession

# Create Spark session
spark = SparkSession.builder.master("local").appName("SparkSQL").getOrCreate()

# Sample Data
data = [
    (1, "Alice", "Google"),
    (2, "Bob", "Amazon"),
    (3, "Charlie", "Microsoft")
]

# Define schema and create DataFrame
columns = ["ID", "NAME", "Company"]
df = spark.createDataFrame(data, schema=columns)

# Register the DataFrame as a temporary SQL view
df.createOrReplaceTempView("employees")

# Use Spark SQL to add a new column "Details" by combining "NAME" and "Company"
result = spark.sql("""
    SELECT *, CONCAT_WS('-', NAME, Company) AS Details
    FROM employees
""")

# Show the result
result.show()
```

# Using SQL CASE WHEN with expr()
### PySpark doesn’t have SQL Like CASE WHEN so in order to use this on PySpark DataFrame withColumn() or select(), you should use expr() function with expression as shown below.

Here, I have used CASE WHEN expression on withColumn() by using expr(), this example updates an existing column gender with the derived values, M for male, F for Female, and unknown for others

```python
data = [("James","M"),("Michael","F"),("Jen","")]
columns = ["name","gender"]
df = spark.createDataFrame(data = data, schema = columns)
df.show()
```

### Replace 'M' with 'Male' & 'F' with 'Female' #else 'Unknown'

```python
from pyspark.sql.functions import expr
```

```python
# Que. Replace 'M' with 'Male' & 'F' with 'Female' #else 'Unknown'
```

```python
#Using CASE WHEN similar to SQL.
df2 = df.withColumn('gender', expr("CASE WHEN gender = 'M' THEN 'MALE'" +
                             "WHEN gender = 'F' THEN 'FEMALE' ELSE 'Unknown' END "))
df2.show()
```

```python
df.withColumn('gender', expr("CASE WHEN gender = 'M' THEN 'MALE'" +
                             "WHEN gender = 'F' THEN 'Female' END"))
```
